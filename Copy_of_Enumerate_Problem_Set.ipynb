```python
'''
1. Given the list
students = ["Ava", "Ben", "Carlos", "Dana"]
Write the code to find a the position of the student that the user enters
EX:
Bot: "What student should I find?"
User: "Ben"
Bot: "Ben is at position 1"
'''
students = ["Ava", "Ben", "Carlos", "Dana"]
student = input('What student should I find')
for ind, val in enumerate(students):
  if val == student:
   print(f'{val} is in position {ind + 1}')
```

```python
students = ["Ava", "Ben", "Carlos", "Dana", 'Edward', 'Fia', 'Gia', 'Henry', 'Isaac']
'''
2. using the above students list,
print out the even numbered students NOT counting the 0th index
'''
for ind, val in enumerate(students):
  if ind % 2 == 0 and ind != 0:
    print(val)
```

```python
'''
3. In the sentence below, find the index of "boring" and replace it with "fun".
'''
sentence = ["Python", "is", "boring", "sometimes"]
for ind, val in enumerate(sentence):
  if val == "boring":
    sentence.pop(ind)
    sentence.insert(ind, "fun")
print(sentence)
```

```python
'''
4. Given this list of numbers, make a new list where only numbers in even index
positions are squared, NOT counting the 0th index
nums = [3, 8, 2, 5, 7, 4]
'''
nums = [3, 8, 2, 5, 7, 4]
newlist = []
for ind, val in enumerate(nums):
  if ind % 2 == 0 and ind != 0:
    newlist.append(val**2)
  else:
    newlist.append(val)
print(newlist)
```
