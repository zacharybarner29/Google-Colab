```python
'''
1: get_smaller takes 2 inputs: a list of numbers and an int
and returns a new list with only the numbers in the list
that are smaller than the input int

ex: get_smaller([1, 10, 2, 5], 3) should return [1,2] because
only 1 and 2 are smaller than 3

'''
def get_smaller(elements, limit):
    new_list = [i for i in elements if i < limit]
    return new_list
print(get_smaller([1, 10, 2, 5], 3))
```

```python
'''
2: get_divisible function takes 2 inputs: a list of numbers and an int
and returns a new list with only the numbers in the list
that are evenly divisible by the input int

ex: get_divisible([10, 11, 12, 13, 14, 15], 5) should return [10,15] because
only 10 and 15 are evenly divisible by 5
'''
def get_divisible(elements, div):
  newlist = [i for i in elements if i % div == 0]
  return newlist
print(get_divisible([10, 11, 12, 13, 14, 15], 5))
```

```python
'''
3: get_shorter function takes 2 inputs: a list of strings and an int
and returns a new list with only the strings in the list
that are shorter than the input int

ex: get_shorter(["hi", "bye", "hello", "goodbye"], 4) should
return ["hi", "bye"] because only "hi" and "bye" are shorter than
length 4
'''
def get_shorter(elements, length):
  newList = [val for val in elements if len(val) < length]
  return newList
print(get_shorter(["hi", "bye", "hello", "goodbye"], 4))
```

```python
'''
4: get_substring function takes 2 inputs: a list of strings and a single substring
and returns a new list with only the strings in the list that contain
the input substring

ex: get_substring(["car", "cats", "a cat", "candles", "caterpillar"], "cat")
should return ["cats", "a cat", "caterpillar"] because only those words
contain the substring "cat"
'''
def get_substrings(elements, sub):
  newList = [val for val in elements if sub in val]
  return newList
print(get_substrings(["car", "cats", "a cat", "candles", "caterpillar"], "cat"))
```
