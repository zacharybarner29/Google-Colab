```python
arcade_tokens = {
    "SkeeBall": 10,
    "Whack-a-Mole": 7,
    "Air Hockey": 15,
    "Dance Dance Revolution": 20,
    "Claw Machine": 5
}
```

```python
#1. The arcade just made some changes.
#Air Hockey is being removed for repairs, and the cost of Whack-a-Mole is going up by 3 tokens.
#Additionally, a new game called "Pinball" costs 8 tokens.
#You need to ...
# - Remove Air Hockey from the dictionary
# - Update the Whack-a-Mole token cost
# - Add Pinball to the dictionary
# - Print the final dictionary

arcade_tokens.pop('Air Hockey')
arcade_tokens['Whack-a-Mole'] += 3 #you can also us a .update()
arcade_tokens['Pinball'] = 8
print(arcade_tokens)
```

```python
#2. Write a program that asks the user to type in a game name and tells them how many tokens it costs.
#If the game doesn't exist in the dictionary, print a friendly message saying it wasn't found instead of crashing.
while True:
  game = input('Enter the name of a game.')
  if game in arcade_tokens:
    print('That is already a game.')
  else:
    tokens = input('How many tokens should it cost?')
    arcade_tokens[game] = tokens
    break
print(arcade_tokens)
```

```python
#3. Write a program that runs a loop and lets the user manage the arcade token list with the following menu:
'''
What would you like to do?
1. See all games and token costs
2. Add a game
3. Remove a game
4. Quit
'''

while True:
  action = int(input('What would you like to do? \n1. See all games and token costs \n2. Add a game \n3. Remove a game \n4. Quit \n'))
  if action == 1:
    print(arcade_tokens)
  elif action == 2:
    while True:
      game = input('Enter the name of a game.')
      if game in arcade_tokens:
        print('That is already a game.')
      else:
        tokens = input('How many tokens should it cost?')
        arcade_tokens[game] = tokens
        break
  elif action == 3:
    arcade_tokens.pop(input('What game would you like to remove?'))
  else:
    break
```
